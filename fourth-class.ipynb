```python
import pandas as pd
```

```python
url = 'https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/base-de-dados/aluguel.csv'
data = pd.read_csv(url, sep=';')
```

```python
data.isnull().sum()
```

```python
data = data.fillna(0)
```

```python
data
```

creating numeric column

```python
data['Valor_por_mes'] = data['Valor'] + data['Condominio']
```

```python
data['Valor_por_ano'] = (data['Valor'] * 12) + (data['Condominio'] * 12) + data['IPTU']
```

creating categoric column

```python
data['Descricao'] = data['Tipo'] + ' em ' + data['Bairro'] + ' com ' + \
                                        data['Quartos'].astype(str) + ' quarto(s) ' + \
                                        ' e ' + data['Vagas'].astype(str) + ' vaga(s) de garagem.'
```

```python
data
```

creating binary column

```python
data['Possui_suite'] = data['Suites'].apply(lambda x: "Sim" if x > 0 else "Não")
```

```python
data.head()
```

```python
data.to_csv('data.csv', index=False)
```

```python
url = 'https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/desafios/alunos.csv'

students = pd.read_csv(url)
```

```python
students = students.fillna(0)
```

```python
students.head()
```

```python
#1

students['Extra_points'] = (students['Notas'] * 1.4) - students['Notas']
```

```python
#2 & 3

students['Final_note'] = students['Notas'] + students['Extra_points']

students['Final_approved'] = students['Final_note'].apply(lambda x: "Yes" if x >= 6 else "Not")
```

```python
students
```

```python
#4

students.query('Aprovado == False & Final_approved == "Yes"')
```
